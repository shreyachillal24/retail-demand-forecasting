```python
import pandas as pd 

df = pd.read_csv(r"C:\Users\shrey\OneDrive\Desktop\retail-demand-forecasting\data\processed\train_cleaned.csv")
df['Date'] = pd.to_datetime(df['Date'])
df.info()
```

```python
 # Step 1: Extracting date features
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Week'] = df['Date'].dt.isocalendar().week
df['DayOfWeek'] = df['Date'].dt.dayofweek
```

```python
df[['Date', 'Year', 'Month', 'Week', 'DayOfWeek']].head()
```

```python
# Step 2 : Lag features 

# Sort the data by store and dept , date 

df = df.sort_values(by=['Store', 'Dept', 'Date'])
```

```python
# Create the lag features 

df['Lag_1'] = df.groupby(['Store', 'Dept'])['Weekly_Sales'].shift(1)  # past week sales
df['Lag_2'] = df.groupby(['Store', 'Dept'])['Weekly_Sales'].shift(2)  # sales from 2 weeks ago 
df['Lag_4'] = df.groupby(['Store', 'Dept'])['Weekly_Sales'].shift(4)  # sales from 4 weeks ago (1montha ago)
```

```python
df[['Store','Dept','Date','Weekly_Sales','Lag_1','Lag_2','Lag_4']].head(10)
```

```python
# step 3 : rolling mean features
df['Rolling_Mean_4'] = df.groupby(['Store', 'Dept'])['Weekly_Sales']\
                        .transform(lambda x: x.shift(1).rolling(4).mean())  # 4 week rolling mean (excluding current week)
```

```python
df['Rolling_Std_4'] = df.groupby(['Store', 'Dept'])['Weekly_Sales']\
                       .transform(lambda x: x.shift(1).rolling(4).std()) # 4 week rolling std(spread)(excluding current week)
```

```python
df[['Store','Dept','Date','Weekly_Sales','Rolling_Mean_4','Rolling_Std_4']].head(10)
```

```python
# Drop the rows with NaN values (due to lag and rolling features) 
df = df.dropna()
```

```python
print(df.shape)
print(df.isnull().sum())
```

```python
# Create the new csv file for the processed data with new features
df.to_csv(r"C:\Users\shrey\OneDrive\Desktop\retail-demand-forecasting\data\processed\train_features.csv", index=False) 
```
