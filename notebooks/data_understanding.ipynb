```python
# step 1 : Understand the data overall and the data types of each column

import pandas as pd 

train = pd.read_csv(r"C:\Users\shrey\OneDrive\Desktop\retail-demand-forecasting\data\raw\train.csv")
test = pd.read_csv(r"C:\Users\shrey\OneDrive\Desktop\retail-demand-forecasting\data\raw\test.csv")
features  = pd.read_csv(r"C:\Users\shrey\OneDrive\Desktop\retail-demand-forecasting\data\raw\features.csv")
stores = pd.read_csv(r"C:\Users\shrey\OneDrive\Desktop\retail-demand-forecasting\data\raw\stores.csv")


print(train.head())

print(test.head())

print(features.head())

print(stores.head())
```

```python
# step 2 : Merge the dataframes together to get a complete picture of the data and to be able to do EDA on the complete data

# Train + features + Stores 

train_merged = pd.merge(train, features, on=['Store', 'Date'], how='left')

train_final = pd.merge(train_merged, stores, on='Store', how='left')

print(train_final.shape)
print(train_final.head())
```

```python
train_final['IsHoliday'] = train_final['IsHoliday_x']
train_final.drop(['IsHoliday_x', 'IsHoliday_y'], axis=1, inplace=True)
```
